### 1. Numpy Array Operations: Temperature Analysis  

```python
# Create an array named temperatures_w1 containing Week 1 temperatures.

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
# Inspect the shape, data type, and the number of elements in the array. 

print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
# Convert the temperatures_w1 from Celsius to Fahrenheit:

fahrenheit = (temperatures_w1 * 9/5) + 32 
print("temperatures in Fahrenheit:", fahrenheit)
```

```python
# Find the maximum, minimum, and mean temperatures for the week:

print("Maximum:", temperatures_w1.max())
print("Minimum:", temperatures_w1.min())
print("Mean:", temperatures_w1.mean())
```

```python
# Extract the temperatures for the first three days of the week:

print(temperatures_w1[:3])
```

```python
# Extract the temperatures for the weekend (last two days):

print(temperatures_w1[-2:])
```

```python
# Extract the temperatures for the middle three days of the week:

print(temperatures_w1[2:5])
```

* Create a 2D Array: 

Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:  

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

```python
#  Inspect and print the shape, data type, and total number of elements:

print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total elements:", temperatures.size)
```

```python
# Extract the temperatures for each week and the weekends (last two days) for both weeks:

week_w1 = temperatures[0]
week_w2 = temperatures[1]

weekend_w1 = temperatures[0, -2:]
weekend_w2 = temperatures[1, -2:]


print("Week 1 temperatures:", week_w1)
print("Week 2 temperatures:", week_w2)
print("Week 1 weekend:", weekend_w1)
print("Week 2 weekend:", weekend_w2)
```

### 2. Pandas Series:   

* Objective: Create a Series containing student marks and perform indexing, filtering and modifications:



```python
# Create a Pandas Series:

marks = pd.Series(
    [95, 92, 89, 85, 80 ],
    index = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print(marks)
```

* Indexing and Slicing:

```python
# Access the first rank using integer position:
print(marks.iloc[0])

# Retrieve the top three ranks using iloc Accessor:
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Retrieve the third rank using iloc Accessor:
print(marks.iloc[2])

# Apply a boolean mask to filter the Series and retrieve the ranks where the marks are >90:
print(marks[marks > 90])
```

* Manipulating Series: 

```python
# Modify the mark of the student with the 1st rank to 100:
marks.loc['Rank1'] = 100

print("After updating Rank1:")
print(marks)

# Remove the entry corresponding to the last rank from the Series:
marks = marks.drop('Rank5') 
print(marks)

# Calculate CGPA by dividing marks by 10:
cgpa = marks / 10

print("Updated marks:\n", marks)
print("CGPA:\n", cgpa)
```

### 3. Pandas DataFrame: 

*Objective: Create a sales transactions dataset and perform data exploration and manipulation:

```python
# Import pandas if you have not already imported it
import pandas as pd

# Creates a Pandas DataFrame:

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400],
})
```

* Data Exploration:

```python
# Display the transactions DataFrame:
display(transactions)

# Display the first 5 rows:
print("First 5 rows:")
display(transactions.head())


# Display the last five rows:
print("Last 5 rows:")
print(transactions.tail())


# Find the shape
print("Shape:",transactions.shape)


# Display the column names:
print("Column names:", transactions.columns.tolist())


# Display the Data types:
print("Data types:")
print(transactions.dtypes)


# Display basic information:
transactions.info()
```

```python
# Display only ProductCategory and Amount:
print(transactions[['ProductCategory', 'Amount']])
```

```python
# Retrieve the last 3 columns of data frames using iloc:
print("Last 3 columns:\n", transactions.iloc[:, -3:])
```

```python
# Filter North region transactions where Amount is greater > 200:
result = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(result)
```

```python
# Find the value counts for the 'ProductCategory' column:
print(transactions['ProductCategory'].value_counts())
```

```python
# Find the unique values in the 'Region' column:
print("Unique regions:", transactions['Region'].unique())
```

```python
# Group by 'Region' and find the mean amount for each region:
print(transactions.groupby('Region')['Amount'].mean())
```

*  Manipulating the DataFrame: 

```python
# Modify the 'Amount' for the transaction with TransactionID 102 to 165:
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

# Display the updated transaction
display(transactions[transactions['TransactionID'] == 102])
```

```python
# Add a Discount column by calculating (10% of Amount):
transactions['Discount'] = transactions['Amount'] * 0.10

display(transactions)
```

```python
#Remove the row with TransactionID 109:
transactions = transactions[
    transactions['TransactionID'] != 109
]

# Display the updated datafrmae:
display(transactions)
```

```python
# Delete the 'Discount' column from the DataFrame:
transactions = transactions.drop(columns=['Discount'])

display(transactions)
```
